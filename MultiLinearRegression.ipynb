```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, r2_score
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("multiple_linear_regression_dataset.csv")
```

```python
print(df.head())

# Prepare features and target
X = df[['age', 'experience']]
y = df['income']

# Split the data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0)

# Initialize and train the model
model = LinearRegression()
model.fit(X_train, y_train)

# Display coefficients and intercept
print("Coefficients:", model.coef_)
print("Intercept:", model.intercept_)

# Make a prediction
prediction = model.predict([[25, 1]])
print("Prediction for age 25 and experience 1:", prediction)

# Plotting
# Plot for 'age'
plt.figure(figsize=(12, 6))

plt.subplot(1, 2, 1)
plt.scatter(X_test['age'], y_test, color='blue', label='Actual values')
plt.scatter(X_test['age'], model.predict(X_test), color='red', label='Predicted values')
plt.title('Age vs Income')
plt.xlabel('Age')
plt.ylabel('Income')
plt.legend()

# Plot for 'experience'
plt.subplot(1, 2, 2)
plt.scatter(X_test['experience'], y_test, color='blue', label='Actual values')
plt.scatter(X_test['experience'], model.predict(X_test), color='red', label='Predicted values')
plt.title('Experience vs Income')
plt.xlabel('Experience')
plt.ylabel('Income')
plt.legend()

plt.tight_layout()
plt.show()

# Test the model's efficacy
y_test_pred = model.predict(X_test)
y_train_pred = model.predict(X_train)

# Print performance metrics
print("Test R² Score:", r2_score(y_test, y_test_pred))
print("Train R² Score:", r2_score(y_train, y_train_pred))

print("Test Mean Squared Error:", mean_squared_error(y_test, y_test_pred))
print("Train Mean Squared Error:", mean_squared_error(y_train, y_train_pred))
```
